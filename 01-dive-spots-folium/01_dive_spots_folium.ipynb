```python
dive_spot_coords = {
    "Punta de la Mona": [36.719435, -3.727315],
    "Roqueo de los 14": [36.721056, -3.728391],
    "Piedra de los Luna": [36.720048, -3.727273],
    "Piedra del Hombre": [36.725500, -3.736306],
    "Cueva de las Palomas": [36.728895, -3.767404],
}
print(dive_spot_coords["Punta de la Mona"])
```

```python
dive_spot_depths = {
    "Punta de la Mona": 49,
    "Roqueo de los 14": 14,
    "Piedra de los Luna": 18,
    "Piedra del Hombre": 10,
    "Cueva de las Palomas": 30,
}
print(dive_spot_depths["Cueva de las Palomas"])
```

```python
for name, coords in dive_spot_coords.items():
    print(f"{name} is at latitude {coords[0]}, longitude {coords[1]}")
```

```python
for name, coords in dive_spot_coords.items():
    depth = dive_spot_depths[name]
    print(f"{name} ({depth} m max) — lat {coords[0]}, lon {coords[1]}")
```

```python
import folium
```

```python
import folium

m = folium.Map(
    location=[36.724, -3.747],
    zoom_start=14,
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}",
    attr="Tiles © Esri, Maxar, Earthstar Geographics and the GIS user community",
)

for name, coords in dive_spot_coords.items():
    depth = dive_spot_depths[name]
    folium.Marker(location=coords, popup=f"{name} — {depth} m").add_to(m)

m
```
